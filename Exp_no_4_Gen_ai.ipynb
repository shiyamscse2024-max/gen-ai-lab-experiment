```python
from transformers import pipeline
```

```python
summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)

article = """
Generative AI refers to artificial intelligence models capable of generating text,
images, audio and videos.
Large Language Models such as GPT and LLaMA are trained on huge text datasets.
These models perform translation, summarization, coding and question answering.
Generative AI is transforming education, healthcare, banking and software development.
"""

summary = summarizer(
    article,
    max_length=45,
    min_length=20,
    do_sample=False
)

print(summary[0]["summary_text"])
```

```python
qa = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad"
)

context = article

question = "What are Large Language Models trained on?"

answer = qa(
    question=question,
    context=context
)

print("Question :",question)
print("Answer :",answer["answer"])
print("Confidence :",answer["score"])
```
