```python
from transformers import pipeline

# Sentiment Analysis
sentiment = pipeline("sentiment-analysis")

reviews = [
    "The new smartphone has an amazing camera and battery life!",
    "The delivery was late and the packaging was damaged."
]

for review in reviews:
    result = sentiment(review)[0]
    print("Review:", review)
    print("Result:", result)
    print()

# Zero-shot Classification
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

document = "The central bank raised interest rates to control inflation."

labels = ["Politics","Economy","Sports","Technology"]

result = classifier(document, labels)

print("Predicted Labels:")
for label, score in zip(result["labels"], result["scores"]):
    print(label, round(score,3))
```
