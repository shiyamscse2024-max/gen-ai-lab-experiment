```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
```

```python
zero_prompt = """
Classify the sentiment of this review as Positive or Negative.

Review:
The product quality is excellent.

Answer:
"""

few_prompt = """
Review: I loved this movie.
Answer: Positive

Review: The food was terrible.
Answer: Negative

Review: The product quality is excellent.
Answer:
"""

cot_prompt = """
Q: A shop had 15 apples.
It sold 6.
Then received 10 more.

Let's think step by step.

15-6 = 9

9+10 = 19

Answer = 19

Q: A library had 120 books.
It lent 45 books.
Then bought 30 books.

Let's think step by step.
"""
```

```python
prompts = [
    ("Zero Shot", zero_prompt),
    ("Few Shot", few_prompt),
    ("Chain of Thought", cot_prompt)
]

for name,p in prompts:
    print("="*60)
    print(name)
    print("="*60)

    result = generator(
        p,
        max_new_tokens=40,
        do_sample=False
    )

    print(result[0]["generated_text"])
    print()
```
